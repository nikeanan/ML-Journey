# Single Variable Linear Regression
This notebook demonstrates simple linear regression fundamentals, model formulation, loss/cost calculation, and plotting predictions using NumPy and Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# x_train: Input size in 1000s sqft
# y_train: Price in $1000s
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])
m = len(x_train)

print(f"x_train = {x_train}")
print(f"y_train = {y_train}")
print(f"Number of training examples (m): {m}")
```

```python
# Scatter plot of dataset
plt.scatter(x_train, y_train, marker='x', c='red', label='Actual Data')
plt.title("Housing Prices")
plt.ylabel('Price ($1000s)')
plt.xlabel('Size (1000 sqft)')
plt.legend()
plt.show()
```

```python
def compute_model_output(x_train, w, b):
    """
    Computes the prediction of a linear model using vectorized NumPy operations.
    Args:
      x_train (ndarray (m,)): Input data
      w, b (scalar)        : Model parameters (weight, bias)
    Returns:
      f_wb (ndarray (m,))  : Model predictions
    """
    return w * x_train + b
```

```python
# Evaluate model predictions with parameters w=200, b=100
w = 200
b = 100
tmp_f_wb = compute_model_output(x_train, w, b)

# Plot model prediction line against actual data points
plt.plot(x_train, tmp_f_wb, c='blue', label='Linear Prediction')
plt.scatter(x_train, y_train, marker='x', c='red', label='Actual Values')
plt.title("Housing Prices Fit")
plt.ylabel('Price ($1000s)')
plt.xlabel('Size (1000 sqft)')
plt.legend()
plt.show()
```
